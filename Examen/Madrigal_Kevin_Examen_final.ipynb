## Conteo paises

```python
# TODO CÓDIGO 2: lista de diccionarios, ciclo, condición y función resumen_region.
# Deben quedar visibles: el conteo y las dos llamadas de prueba de resumen_region.

registros = df.to_dict(orient='records')

conteo_puntuacion_bienestar = 0
for registro in registros:
    if registro['puntuacion_bienestar'] >= 6.0:
        conteo_puntuacion_bienestar += 1

print(f"La cantidad de paises con una puntuacion de bienestar mayor o igual a 6.0 es: {conteo_puntuacion_bienestar}")
        
```

## Funcion resumen_region()

```python
def resumen_region(registros, region):
    
    registros_region = [registro for registro in registros if registro['region'] == region]
    if registros_region:
        suma_bienestar = 0
        cantidad_paises = len(registros_region)
        
        for registro in registros_region:
            suma_bienestar += registro['puntuacion_bienestar']
            
        promedio_bienestar = suma_bienestar / cantidad_paises
        
        return region, cantidad_paises, promedio_bienestar
    
resultado = resumen_region(registros, 'Europa Occidental')

if resultado is not None:
    print(f"Region: {resultado[0]}\nCantidad de paises: {resultado[1]}\nPromedio de Bienestar: {resultado[2]}")
else:
    print(f"La region que busca no existe")
            
```

```python
resultado = resumen_region(registros, 'Centroamerica')

if resultado is not None:
    print(f"Region: {resultado[0]}\nCantidad de paises: {resultado[1]}\nPromedio de Bienestar: {resultado[2]}")
else:
    print(f"{resultado}: La region que busca no esta registrada en la base de datos")
```

## Explicacion

Se usa una comprension de listas para ingresar unicamente los registros del diccionario que pertenecen a la region que se esta buscando en una lista.
El if evalua si la lista contiene elementos (que seria True y si no False y no entra al bucle). Si la lista esta vacia significa que no existe la region buscada.
Si se entra al bucle se hacen todos los calculos necesarios que se piden y retorna explicitamente lo que se pide
Si no entra al bucle la funcion retorna por defecto None

De modo que si existe la region se hacen los calculos y se devuelven los resultados y si no existe se devuelve None.

```python
resumen_por_region = df.groupby('region').agg({'puntuacion_bienestar': 'mean', 'pais': 'count'}).reset_index().sort_values(by='puntuacion_bienestar', ascending=False).rename(columns={'puntuacion_bienestar': 'promedio_bienestar', 'pais': 'cantidad_paises'})
resumen_por_region
```
