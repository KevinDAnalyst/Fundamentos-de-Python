```python
#EXTRACCION
import pandas as pd # Importar libreria de pandas

url_bccr = "https://gee.bccr.fi.cr/IndicadoresEconomicos/Cuadros/frmConsultaTCVentanilla.aspx"

tablas_web = pd.read_html(url_bccr, encoding='UTF-8', decimal=",", thousands=".") # read_html() devuelve el contenido del sitio
print("Tipo de datos de tabla web:", type(tablas_web))
print("Numero de tablas encontradas:", len(tablas_web))

df = tablas_web[2].copy()
# Analisis
print("")
print(df.info())
print("")

print("\n---DIMENSIONES---")
print(df.shape)
df.head()
```

## 🧹 Limpieza y Transformación (La Evolución de Nuestro ETL Robusto)

Recordarás la importancia de la limpieza y saneamiento de datos del Notebook 06. Pandas no solo lo hace más fácil, sino que también nos da herramientas mucho más potentes para manejar los datos sucios del mundo real (valores faltantes, errores de formato, etc.).

```python
# Eliminar Valores Nulos
df.dropna()
```

```python
# Rellenar valores nulos
df.fillna("No Valor")
df.head(3)
```

```python
# Reemplazar nulos con el ultimo valor hasta que aparezca uno diferente
# df = df.ffill()
df.ffill(inplace=True)
```

```python
# Encabezados
# Extraer la primera fila
primera_fila = df.iloc[0]

# ASIGNAR al DF la primera fila como columnas
df.columns = primera_fila
df.head(3)

# Eliminar la serie (fila) 0, la duplicada
df.drop(0, inplace=True)
df.tail(3)
```

```python
# Renombrar columnas
df.rename(columns={"Tipo de Entidad": "TIPO", "Entidad Autorizada": "ENTIDAD", "Compra": "COMPRA", "Venta": "VENTA", "Diferencial Cambiario": "DIFERENCIAL", "Última Actualización": "FECHA"}, inplace=True)
df.head()
```

```python
# Eliminar duplicados
df.drop_duplicates(inplace=True)
df.tail()
```

```python
# Tipado de series en el DataFrame (Conversion)
columnas_float = ["COMPRA", "VENTA", "DIFERENCIAL"]
#df[columnas_float].apply(pd.to_numeric, inplace=True)
df[columnas_float] = df[columnas_float].astype(float, errors='ignore')

df["FECHA"].apply(pd.to_datetime, dayfirst=True, errors='coerce')
df.info()
```

```python
# Carlos gana en $, quiere saber que entidad paga mas
tipo_compra_max = df["COMPRA"].max()


filtro = df["COMPRA"] == tipo_compra_max
lista_entidades = df[filtro]
print(f"La entidad que mas paga es: \n{lista_entidades['ENTIDAD']} con un valor de: {lista_entidades['COMPRA']}")
```

```python
# Carlos gana en dolares, quieres saber que entidad paga mas
tipo_compra_max = df['COMPRA'].max()
print(tipo_compra_max)
#lista_entidades = df[df['COMPRA'] == tipo_compra_max]['ENTIDAD'].tolist()
filtro = df['COMPRA'] == tipo_compra_max
lista_entidades = df[filtro]
print(f"La entidad que paga mas: {lista_entidades['ENTIDAD'].values} con un valor de: {lista_entidades['COMPRA']}")
#
```
