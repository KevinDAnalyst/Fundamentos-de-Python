# Caso 1: Control de ventas diarias de una soda

## Problema
Una soda ubicada en San Pedro registra sus ventas de lunes a viernes. El programa debe calcular el total semanal, el promedio diario y clasificar el resultado según una meta.

## Entrada / datos
Use esta lista base:

```python
ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000
```

## Proceso esperado

- Calcular total semanal.
- Calcular promedio diario.
- Comparar el total con la meta.
- Clasificar:
  - `Meta alcanzada` si el total es mayor o igual a la meta.
  - `Cerca de la meta` si el total alcanza al menos el 80% de la meta.
  - `Requiere atención` si no alcanza el 80%.

## Salida esperada
Debe mostrar total, promedio, porcentaje de cumplimiento y estado final.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[85000, 92000, 78000, 110000, 97000]`, meta `450000` | Meta alcanzada | Meta alcanzada (Total: ₡462,000, Cumplimiento: 102.67%) |
| Caso límite | total exactamente igual a la meta ([90000, 90000, 90000, 90000, 90000], meta 450000) | Meta alcanzada | Meta alcanzada (Total: ₡450,000, Cumplimiento: 100.00%) |
| Caso especial | total menor al 80% de la meta ([50000, 40000, 60000, 55000, 45000], meta 450000) | Requiere atención | Requiere atención (Total: ₡250,000, Cumplimiento: 55.56%) |

````python
# Caso 1 - Solución Desarrollada
ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000

# Cálculos financieros
total_semanal = sum(ventas_soda)
promedio_diario = total_semanal / len(ventas_soda)
porcentaje_cumplimiento = (total_semanal / meta_semanal) * 100

# Clasificación según el porcentaje e indicadores de metas
if total_semanal >= meta_semanal:
    estado_final = "Meta alcanzada"
elif total_semanal >= (meta_semanal * 0.80):
    estado_final = "Cerca de la meta"
else:
    estado_final = "Requiere atención"

# Reporte de Resultados
print("======= REPORTE DE VENTAS SODA SAN PEDRO =======")
print(f"Total de ventas semanal: ₡{total_semanal:,}")
print(f"Promedio de ventas diario: ₡{promedio_diario:,.2f}")
print(f"Porcentaje de cumplimiento: {porcentaje_cumplimiento:.2f}%")
print(f"Estado de rendimiento comercial: {estado_final}")
print("================================================")
````

# Caso 2: Asistencia a una actividad cultural

## Problema
Una actividad cultural en Costa Rica registra asistencia por bloques horarios. El programa debe calcular la asistencia promedio y clasificar el nivel de ocupación.

## Entrada / datos
Use esta lista base:

```python
asistencias = [120, 180, 210, 250, 230]
capacidad = 300
```

## Proceso esperado

- Recorrer la lista con un ciclo `for`.
- Calcular asistencia total y promedio.
- Contar cuántos bloques superaron el 70% de capacidad.
- Clasificar el promedio de ocupación:
  - `Alta ocupación` si el promedio es mayor o igual al 80% de la capacidad.
  - `Ocupación media` si el promedio es mayor o igual al 50%.
  - `Baja ocupación` si el promedio es menor al 50%.

## Salida esperada
Debe mostrar total de asistentes, promedio, bloques sobre 70% y estado de ocupación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[120, 180, 210, 250, 230]`, capacidad `300` | Ocupación media o alta según cálculo | Ocupación media (Promedio: 198.0, Bloques >70%: 2) |
| Caso límite | promedio exactamente 80% de capacidad ([240, 240, 240, 240, 240], cap. 300) | Alta ocupación | Alta ocupación (Promedio: 240.0, es exactamente el 80%) |
| Caso especial | lista con asistencias muy bajas ([40, 30, 50, 60, 45], cap. 300) | Baja ocupación | Baja ocupación (Promedio: 45.0, Bloques >70%: 0) |

````python
# Caso 2 - Solución Desarrollada
asistencias = [120, 180, 210, 250, 230]
capacidad = 300


total_asistentes = 0
bloques_criticos = 0
umbral_critico_bloque = capacidad * 0.70


for asistencia in asistencias:
    total_asistentes += asistencia
    if asistencia > umbral_critico_bloque:
        bloques_criticos += 1

promedio_asistencia = total_asistentes / len(asistencias)

if promedio_asistencia >= (capacidad * 0.80):
    estado_ocupacion = "Alta ocupación"
elif promedio_asistencia >= (capacidad * 0.50):
    estado_ocupacion = "Ocupación media"
else:
    estado_ocupacion = "Baja ocupación"

# Salida 
print("======= REPORTE OPERATIVO DE EVENTO CULTURAL =======")
print(f"Total general de asistentes: {total_asistentes} personas")
print(f"Asistencia promedio por bloque: {promedio_asistencia:.1f} personas")
print(f"Bloques que superaron el 70% de capacidad (>210): {bloques_criticos}")
print(f"Nivel de ocupación promedio general: {estado_ocupacion}")
print("====================================================")
````

# Caso 3: Precios de productos de feria

## Problema
Una persona compara precios de un mismo producto en diferentes ferias del agricultor. El programa debe analizar los precios y detectar si hay mucha variación.

## Entrada / datos
Use esta lista base:

```python
producto = "Tomate"
precios = [900, 850, 1000, 950, 1200]
```

## Proceso esperado

- Calcular precio promedio.
- Identificar precio mínimo y máximo.
- Calcular rango de precios: `máximo - mínimo`.
- Clasificar la variación:
  - `Variación alta` si el rango es mayor o igual a 300.
  - `Variación moderada` si el rango es mayor o igual a 150.
  - `Variación baja` si el rango es menor a 150.
- Usar un ciclo para contar cuántos precios están por encima del promedio.

## Salida esperada
Debe mostrar producto, promedio, mínimo, máximo, rango, precios sobre promedio y clasificación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[900, 850, 1000, 950, 1200]` | Variación alta | Variación alta (Rango: ₡350, Precios sobre promedio: 2) |
| Caso límite | rango exactamente `300` ([900, 950, 1000, 1100, 1200]) | Variación alta | Variación alta (Rango: ₡300 exactamente) |
| Caso especial | todos los precios iguales ([1000, 1000, 1000, 1000, 1000]) | Variación baja | Variación baja (Rango: ₡0, Precios sobre promedio: 0) |

````python
# Caso 3 - Solución Desarrollada
producto = "Tomate"
precios = [900, 850, 1000, 950, 1200]

# Computar indicadores analíticos
precio_promedio = sum(precios) / len(precios)
precio_minimo = min(precios)
precio_maximo = max(precios)
rango_precios = precio_maximo - precio_minimo

# Evaluar el nivel de dispersión y variación de mercado
if rango_precios >= 300:
    clasificacion_variacion = "Variación alta"
elif rango_precios >= 150:
    clasificacion_variacion = "Variación moderada"
else:
    clasificacion_variacion = "Variación baja"

# Bucle de conteo para elementos que superan el punto medio
establecimientos_sobre_promedio = 0
for precio in precios:
    if precio > precio_promedio:
        establecimientos_sobre_promedio += 1

# Presentación formal de resultados analíticos
print(f"======= ANÁLISIS DE MERCADO: {producto.upper()} =======")
print(f"Precio promedio calculado: ₡{precio_promedio:,.2f}")
print(f"Precio mínimo de mercado: ₡{precio_minimo:,}")
print(f"Precio máximo de mercado: ₡{precio_maximo:,}")
print(f"Rango absoluto de variación: ₡{rango_precios:,}")
print(f"Puntos de venta con precios por encima del promedio: {establecimientos_sobre_promedio}")
print(f"Clasificación de estabilidad de precios: {clasificacion_variacion}")
print("=======================================================")
````

# Caso 4: Reto integrador - Red EmprendeCR

## Problema
Una red de emprendimientos costarricenses desea analizar el desempeño semanal de varias sedes. Cada sede tiene nombre, provincia, tipo de negocio, ventas diarias y meta semanal.

Este reto integra aprendizajes de las semanas 1 a 7:

- Semana 1: comprensión del problema y algoritmo.
- Semana 2: variables y entrada-proceso-salida.
- Semana 3: decisiones con condicionales.
- Semana 4: validación y código limpio.
- Semana 5: indicadores para análisis de datos.
- Semana 6: listas y ciclos.
- Semana 7: listas de diccionarios, funciones y elección de estructuras.

## Datos base

Use esta lista de diccionarios:

```python
sedes = [
    {
        "nombre": "Soda San Pedro",
        "provincia": "San José",
        "tipo": "Alimentación",
        "ventas": [85000, 92000, 78000, 110000, 97000],
        "meta": 450000,
    },
    {
        "nombre": "Panadería Cartago",
        "provincia": "Cartago",
        "tipo": "Panadería",
        "ventas": [65000, 70000, 72000, 69000, 75000],
        "meta": 380000,
    },
    {
        "nombre": "Cafetería Liberia",
        "provincia": "Guanacaste",
        "tipo": "Cafetería",
        "ventas": [98000, 105000, 99000, 120000, 115000],
        "meta": 500000,
    },
]
```

## Requisitos mínimos

1. Crear una función `calcular_total(ventas)`.
2. Crear una función `calcular_promedio(ventas)`.
3. Crear una función `clasificar_sede(total, meta)`.
4. Recorrer todas las sedes con un ciclo `for`.
5. Imprimir un reporte por sede con:
   - nombre;
   - provincia;
   - total semanal;
   - promedio diario;
   - porcentaje de cumplimiento;
   - estado.
6. Crear un `set` con las provincias presentes.
7. Crear una lista de tuplas `(nombre, total)` para guardar el ranking base.
8. Identificar la sede con mayor venta total.

## Clasificación requerida

- `Meta alcanzada`: total mayor o igual a la meta.
- `Cerca de la meta`: total mayor o igual al 80% de la meta.
- `Requiere atención`: total menor al 80% de la meta.

## Salida esperada
El reporte debe ser claro y comprensible para una persona que no está viendo el código.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | datos base | Reporte para las 3 sedes | Reporte completo impreso sin errores. Líder: Cafetería Liberia (₡537,000) |
| Caso límite | una sede con total igual a la meta | Meta alcanzada | Cumplimiento del 100.00% exacto asignando Meta alcanzada |
| Caso especial | una sede con ventas muy bajas (Sede con ventas de cero colones en sus registros diarios) | Requiere atención | Cumplimiento del 0.00% asignando requiere atencion sin corromper el script |

````python
# Caso 4 - Reto Integrador
sedes = [
    {
        "nombre": "Soda San Pedro",
        "provincia": "San José",
        "tipo": "Alimentación",
        "ventas": [85000, 92000, 78000, 110000, 97000],
        "meta": 450000,
    },
    {
        "nombre": "Panadería Cartago",
        "provincia": "Cartago",
        "tipo": "Panadería",
        "ventas": [65000, 70000, 72000, 69000, 75000],
        "meta": 380000,
    },
    {
        "nombre": "Cafetería Liberia",
        "provincia": "Guanacaste",
        "tipo": "Cafetería",
        "ventas": [98000, 105000, 99000, 120000, 115000],
        "meta": 500000,
    },
]

# 1. Defina las funciones del sistema corporativo
def calcular_total(ventas):
    """Calcula y retorna la sumatoria completa de una lista de ventas."""
    return sum(ventas)

def calcular_promedio(ventas):
    """Calcula y retorna el promedio diario de una lista de ventas."""
    if len(ventas) == 0:
        return 0
    return sum(ventas) / len(ventas)

def clasificar_sede(total, meta):
    """Determina la etiqueta cualitativa de desempeño basado en la meta comercial."""
    if total >= meta:
        return "Meta alcanzada"
    elif total >= (meta * 0.80):
        return "Cerca de la meta"
    else:
        return "Requiere atención"


# Inicialización de colecciones de datos y variables de control analítico
provincias_presentes = set()
ranking_base = []
mejor_sede_nombre = ""
mejor_venta_monto = -1

# 2. Recorra las sedes y genere el reporte individual detallado
print("=====================================================================")
print("             REPORTES DE RENDIMIENTO INDIVIDUAL - RED EMPRENDECR      ")
print("=====================================================================")

for sede in sedes:
    # Extracción segura de atributos del diccionario
    nombre = sede["nombre"]
    provincia = sede["provincia"]
    tipo = sede["tipo"]
    ventas_lista = sede["ventas"]
    meta_establecida = sede["meta"]
    
    # Invocación funcional para derivación de métricas
    total_semanal = calcular_total(ventas_lista)
    promedio_diario = calcular_promedio(ventas_lista)
    porcentaje_cumplimiento = (total_semanal / meta_establecida) * 100
    estado_desempeno = clasificar_sede(total_semanal, meta_establecida)
    
    # Inserción y actualización de colecciones nativas de control
    provincias_presentes.add(provincia)
    ranking_base.append((nombre, total_semanal))
    
    # Búsqueda de valor máximo para identificar la mejor sede
    if total_semanal > mejor_venta_monto:
        mejor_venta_monto = total_semanal
        mejor_sede_nombre = nombre
        
    # Impresión de la ficha del establecimiento
    print(f"Establecimiento: {nombre} ({tipo})")
    print(f"  -> Ubicación geográfica: {provincia}, Costa Rica")
    print(f"  -> Facturación total semanal: ₡{total_semanal:,}")
    print(f"  -> Promedio de ingreso diario: ₡{promedio_diario:,.2f}")
    print(f"  -> Grado de cumplimiento real: {porcentaje_cumplimiento:.2f}%")
    print(f"  -> Clasificación operativa:    [{estado_desempeno}]")
    print("-" * 69)

# 3. Muestre provincias presentes, ranking base y mejor sede (Auditoría Global)
print("\n=====================================================================")
print("               INFORME MACRO DE AUDITORÍA - RED EMPRENDECR           ")
print("=====================================================================")
print(f"Provincias con cobertura activa (Set): {provincias_presentes}")
print(f"Historial consolidado de ranking (Tuplas): {ranking_base}")
print(f"Establecimiento líder de la red comercial: {mejor_sede_nombre} con ₡{mejor_venta_monto:,}")
print("=====================================================================")
````
