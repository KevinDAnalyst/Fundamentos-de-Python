## Nivel intermedio - 6. Cálculos básicos

Calcule totales y agregue una columna nueva.


```python
# TODO: calcule el total de nacimientos
# total_nacimientos = ...


total_nacimientos = df['Nacimientos'].sum()


# TODO: calcule el total de defunciones
# total_defunciones = ...
total_defunciones = df['Defunciones'].sum()

# TODO: cree la columna Saldo Poblacional
# df["Saldo Poblacional"] = ...
df['Saldo_Poblacional'] = total_nacimientos - total_defunciones

# TODO: muestre total_nacimientos, total_defunciones y df.head()
print("Total de nacimientos:", total_nacimientos)
print("Total de defunciones:", total_defunciones)
print("Saldo Poblacional:", df['Saldo_Poblacional'].sum())
df.head()
```

## Nivel intermedio - 7. Gráfico de líneas: nacimientos por año

Cree un gráfico de líneas con `Año` en el eje X y `Nacimientos` en el eje Y.


```python
# TODO: cree una figura con figsize=(10, 4)
plt.figure(figsize=(10, 4))
# TODO: grafique Año vs Nacimientos
plt.plot(df["Año"], df["Nacimientos"], label="Nacimientos", color="#5893CA")
# TODO: agregue título, etiqueta X, etiqueta Y y plt.tight_layout()
plt.title("Nacimientos por Año", fontweight="bold", fontsize=14)
plt.xlabel("Año")
plt.ylabel("Nacimientos")
plt.tight_layout()
# TODO: muestre el gráfico con plt.show()
plt.show()
```

**Pregunta de interpretación:** ¿los nacimientos muestran una tendencia general de aumento, disminución o estabilidad?


## Nivel intermedio - 8. Comparación de nacimientos y defunciones

Cree un gráfico con dos líneas y una leyenda.


```python

fig, ax = plt.subplots(figsize=(12, 4))

plt.plot(df["Año"], df["Nacimientos"], label="Nacimientos", color="#5893CA")
plt.plot(df["Año"], df["Defunciones"], label="Defunciones", color="#F28E2B")

plt.title("Nacimientos y Defunciones por Año", fontweight="bold", fontsize=14)
plt.xlabel("Año")
plt.ylabel("Cantidad")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# TODO: cree el gráfico final
# TODO: ¿Cuáles son los 10 años con más nacimientos?
# Requisitos: título, etiquetas, tamaño adecuado e interpretación escrita debajo

df_top10_nacimientos = df.nlargest(10, "Nacimientos")
df_top10_nacimientos = df_top10_nacimientos.sort_values("Año")  # Ordenar por año para una mejor visualización

```

```python
plt.figure(figsize=(10, 6))
plt.plot(df_top10_nacimientos["Año"], df_top10_nacimientos["Nacimientos"], marker='o', color="#5893CA")
plt.title("Top 10 Años con Más Nacimientos", fontsize=16, fontweight='bold')
plt.xlabel("Año")
plt.ylabel("Cantidad de Nacimientos")

plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

**Interpretación final:** escriba 3 o 4 frases explicando qué muestra su gráfico y qué conclusión obtiene.


Este gráfico de líneas muestra la evolución de la cantidad de nacimientos a lo largo de los 10 años con mayor natalidad registrados en tus datos. El eje horizontal representa los años (ordenados cronológicamente desde 1985 hasta 1995), mientras que el eje vertical mide la cantidad de nacimientos anuales. La línea permite identificar un pico de natalidad al inicio del periodo en 1985, seguido de una fuerte caída hacia 1987 y una recuperación temporal en 1989. Como conclusión principal, se observa una clara tendencia general a la baja en la natalidad a largo plazo, estabilizándose en sus niveles mínimos hacia mediados de la década de 1990.
